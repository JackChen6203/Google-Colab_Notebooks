```python
import socket, urllib.request
for host in ('github.com', 'pypi.org', 'www.googleapis.com'):
    socket.getaddrinfo(host, 443, type=socket.SOCK_STREAM)
    request = urllib.request.Request('https://' + host, headers={'User-Agent': 'Kaggle-PromptDB-network-diagnostic'})
    try:
        with urllib.request.urlopen(request, timeout=20) as response:
            print(host, 'DNS OK, HTTPS', response.status)
    except urllib.error.HTTPError as error:
        print(host, 'DNS OK, HTTPS response', error.code)
```
